### Tunnels Setup

**WARNING:** Ngrok is currently the default tunnel option. Keep in mind that all free tunneling methods, including Ngrok, have limitations and may stop working unexpectedly from one day to the next.

Select the Tunneling Service to Generate the Public URL:

- Gradio + LocalTunnel: This will use Gradio for the Applio UI and LocalTunnel for the Tensorboard and Filebrowser. Select it in Tunnel, run the cell, wait for the Local URL to appear, open the Gradio Public URL for the Applio UI, instead for the Tensorboard and Filebrowser you need to copy the LocalTunnel Password and paste it in Tunnel Password of the LocalTunnel Tunnel Public URL. 

- Ngrok: Select it in Tunnel, get the Ngrok Tunnel Authtoken here: https://dashboard.ngrok.com/tunnels/authtokens/new, put it in ngrok_tunnel_authtoken, run the, wait for the Local URL to appear and click on the Ngrok Tunnel Public URL which is above.

- LocalTunnel: Select it in Tunnel, run the cell, wait for the Local URL to appear, copy the LocalTunnel Password displayed under the public link and paste it in Tunnel Password of the LocalTunnel Tunnel Public URL which is above.

- Horizon: Select it in Tunnel, get the Horizon ID here: https://hrzn.run/dashboard/, login, on the 2nd step, it shows an hrzn login YOUR_ID, you need to copy that id and put it in horizon_id. Then run the cell, you will get an 'HORIZON: Authorize at https://hrzn.run/dashboard/settings/cli-token-requests/YOUR_ID', click it, do Approve. At the end, run the cell, wait for the Local URL to appear and click on the Horizon Tunnel Public URL which is above.

```python
import os
import re
import subprocess
import time
import urllib.request
from collections import deque
from IPython.display import clear_output

Tunnel = "Gradio + LocalTunnel" #@param ["Gradio + LocalTunnel", "Ngrok", "LocalTunnel", "Horizon"]
ngrok_authtoken = "" #@param {type:"string"}
horizon_id = "" #@param {type:"string"}

%cd /kaggle/working/program_ml
os.system("filebrowser -r /kaggle -p 9876 > /dev/null 2>&1 &")
%load_ext tensorboard
%tensorboard --logdir logs --port 8077

share_option = ""
session_links = []

if Tunnel == "Gradio + LocalTunnel":
  print("Using Gradio's built-in tunneling for Applio UI and LocalTunnel for Tensorboard and Filebrowser")
  share_option = "--share"
  !npm install -g localtunnel
  import urllib
  with open('t.txt', 'w') as file:
    file.write('')
  get_ipython().system_raw('lt --port 8077 >> t.txt 2>&1 &')
  time.sleep(7)
  endpoint_ip = urllib.request.urlopen('https://ipv4.icanhazip.com').read().decode('utf8').strip("\n")
  with open('t.txt', 'r') as file:
    t_tunnel = file.read().replace("your url is: ", "").strip()

  with open('f.txt', 'w') as file:
    file.write('')
  get_ipython().system_raw('lt --port 9876 >> f.txt 2>&1 &')
  time.sleep(7)
  endpoint_ip = urllib.request.urlopen('https://ipv4.icanhazip.com').read().decode('utf8').strip("\n")
  with open('f.txt', 'r') as file:
    f_tunnel = file.read().replace("your url is: ", "").strip()

  session_links = [
    ("TensorBoard ", t_tunnel),
    ("FileBrowser ", f_tunnel),
    ("LT Password ", endpoint_ip),
  ]
elif Tunnel == "Ngrok":
  !pip install -q pyngrok
  from pyngrok import ngrok
  ngrok.set_auth_token(ngrok_authtoken)
  p_tunnel = ngrok.connect(6969)
  t_tunnel = ngrok.connect(8077)
  f_tunnel = ngrok.connect(9876)
  session_links = [
    ("Applio UI   ", p_tunnel.public_url),
    ("TensorBoard ", t_tunnel.public_url),
    ("FileBrowser ", f_tunnel.public_url),
  ]
elif Tunnel == "LocalTunnel":
  !npm install -g localtunnel
  import urllib
  with open('p.txt', 'w') as file:
    file.write('')
  get_ipython().system_raw('lt --port 6969 >> p.txt 2>&1 &')
  time.sleep(7)
  endpoint_ip = urllib.request.urlopen('https://ipv4.icanhazip.com').read().decode('utf8').strip("\n")
  with open('p.txt', 'r') as file:
    p_tunnel = file.read().replace("your url is: ", "").strip()

  with open('t.txt', 'w') as file:
    file.write('')
  get_ipython().system_raw('lt --port 8077 >> t.txt 2>&1 &')
  time.sleep(7)
  endpoint_ip = urllib.request.urlopen('https://ipv4.icanhazip.com').read().decode('utf8').strip("\n")
  with open('t.txt', 'r') as file:
    t_tunnel = file.read().replace("your url is: ", "").strip()

  with open('f.txt', 'w') as file:
    file.write('')
  get_ipython().system_raw('lt --port 9876 >> f.txt 2>&1 &')
  time.sleep(7)
  endpoint_ip = urllib.request.urlopen('https://ipv4.icanhazip.com').read().decode('utf8').strip("\n")
  with open('f.txt', 'r') as file:
    f_tunnel = file.read().replace("your url is: ", "").strip()

  session_links = [
    ("Applio UI   ", p_tunnel),
    ("TensorBoard ", t_tunnel),
    ("FileBrowser ", f_tunnel),
    ("LT Password ", endpoint_ip),
  ]
elif Tunnel == "Horizon":
  !npm install -g @hrzn/cli
  !hrzn login $horizon_id
  with open('p.txt', 'w') as file:
    file.write('')
  get_ipython().system_raw('hrzn tunnel http://localhost:6969 >> p.txt 2>&1 &')
  time.sleep(7)
  p_tunnel = !grep -oE "https://[a-zA-Z0-9.-]+\.hrzn\.run" p.txt
  p_tunnel = p_tunnel[0]

  with open('t.txt', 'w') as file:
    file.write('')
  get_ipython().system_raw('hrzn tunnel http://localhost:8077 >> t.txt 2>&1 &')
  time.sleep(7)
  t_tunnel = !grep -oE "https://[a-zA-Z0-9.-]+\.hrzn\.run" t.txt
  t_tunnel = t_tunnel[0]

  with open('f.txt', 'w') as file:
    file.write('')
  get_ipython().system_raw('hrzn tunnel http://localhost:9876 >> f.txt 2>&1 &')
  time.sleep(7)
  f_tunnel = !grep -oE "https://[a-zA-Z0-9.-]+\.hrzn\.run" f.txt
  f_tunnel = f_tunnel[0]

  session_links = [
    ("Applio UI   ", p_tunnel),
    ("TensorBoard ", t_tunnel),
    ("FileBrowser ", f_tunnel),
  ]

# --- Clean & Optimized Live Kaggle Console Monitor ---
BOLD = "\033[1m"
RESET = "\033[0m"
ANSI_RE = re.compile(r"\x1B\[[0-?]*[ -/]*[@-~]")
EPOCH_RE = re.compile(
  r"epoch=(\d+)\s*\|\s*step=(\d+).*?training_speed=(\d+:\d{2}:\d{2})"
)
PUBLIC_URL_RE = re.compile(r"Running on public URL:\s*(https?://\S+)")
LOCAL_URL_RE = re.compile(r"Running on local URL:\s*(https?://\S+)")
NOISE_RE = re.compile(
  r"(\d+%\||it/s\]|s/it\]|[━█▓▒░]{3,}|"
  r"FutureWarning|UserWarning|DeprecationWarning|RuntimeWarning|pkg_resources|"
  r"warnings\.warn|oneDNN|tensorflow/core|computation_placer|numa_node|"
  r"Successfully opened dynamic library|weights_only|gradio\.analytics|"
  r"HTTP Request:|Checking out|libpng warning|ALSA|jack server|To create a public link)",
  re.IGNORECASE,
)
MILESTONE_RE = re.compile(
  r"(Starting preprocess|Preprocess completed|preprocessed successfully|"
  r"Starting pitch extraction|Pitch extraction completed|"
  r"Starting embedding extraction|Embedding extraction completed|"
  r"Saved index file|Starting training|Saving checkpoint|Saving model|"
  r"Training finished|Training completed|Index file.*generated|"
  r"Using .* vocoder|Using .* for training)",
  re.IGNORECASE,
)
ERROR_START_RE = re.compile(
  r"(Traceback \(most recent call last\)|RuntimeError:|ValueError:|"
  r"FileNotFoundError:|OSError:|ModuleNotFoundError:|ImportError:|"
  r"CUDA out of memory|OutOfMemoryError|Exception:|\bERROR:)",
)

milestones = deque(maxlen=6)
errors = deque(maxlen=8)
epoch_rows = deque(maxlen=15)
seen_epochs = set()
tb_lines_seen = 0

def upsert_link(label, value):
  for idx, (k, _) in enumerate(session_links):
    if k.strip() == label.strip():
      session_links[idx] = (label, value)
      return
  session_links.insert(0, (label, value))

def render_monitor(status="Running"):
  w_speed = max(9, max((len(r[0]) for r in epoch_rows), default=0))
  w_step = max(8, max((len(r[1]) for r in epoch_rows), default=0))
  w_epoch = max(6, max((len(r[2]) for r in epoch_rows), default=0))

  lines = [
    f"{BOLD}NoRegretsRVC Kaggle Live Monitor{RESET}  [{status}]",
    "=" * 54,
  ]
  for label, val in session_links:
    if val:
      lines.append(f"{label}: {val}")
  lines.append("-" * 54)

  if milestones:
    for m in milestones:
      lines.append(f"[Stage] {m}")
    lines.append("")

  if epoch_rows:
    lines.append(f"{'Speed':<{w_speed}} | {'Step':<{w_step}} | {BOLD}{'Epoch':<{w_epoch}}{RESET}")
    lines.append(f"{'-' * w_speed}-+-{'-' * w_step}-+-{'-' * w_epoch}")
    for spd, stp, ep in epoch_rows:
      lines.append(f"{spd:<{w_speed}} | {stp:<{w_step}} | {BOLD}{ep:<{w_epoch}}{RESET}")
  elif not errors:
    lines.append("Waiting for training epochs...")

  if errors:
    lines.append("")
    lines.append(f"{BOLD}Errors:{RESET}")
    for err in errors:
      lines.append(f"  {err}")

  clear_output(wait=True)
  print("\n".join(lines), flush=True)

env = os.environ.copy()
env.update({
  "PYTHONWARNINGS": "ignore",
  "PYTHONUNBUFFERED": "1",
  "TF_CPP_MIN_LOG_LEVEL": "3",
  "TORCH_CPP_LOG_LEVEL": "ERROR",
  "TOKENIZERS_PARALLELISM": "false",
  "TQDM_DISABLE": "1",
})

cmd = ["/kaggle/.venv/bin/python", "-u", "app.py", "--client", "--host", "0.0.0.0", "--port", "6969"]
if share_option:
  cmd.append(share_option)

render_monitor("Starting WebUI")
proc = subprocess.Popen(
  cmd,
  stdout=subprocess.PIPE,
  stderr=subprocess.STDOUT,
  text=True,
  encoding="utf-8",
  errors="replace",
  bufsize=1,
  env=env,
)

try:
  assert proc.stdout is not None
  for raw_line in proc.stdout:
    line = ANSI_RE.sub("", raw_line).strip()
    if not line:
      continue

    pub_m = PUBLIC_URL_RE.search(line)
    if pub_m:
      upsert_link("Applio UI   ", pub_m.group(1))
      render_monitor("Ready")
      continue

    loc_m = LOCAL_URL_RE.search(line)
    if loc_m and not any(k.strip() == "Applio UI" for k, _ in session_links):
      upsert_link("Local UI    ", loc_m.group(1))
      render_monitor("Ready")
      continue

    ep_m = EPOCH_RE.search(line)
    if ep_m:
      ep_num, stp_num, spd_str = ep_m.group(1), ep_m.group(2), ep_m.group(3)
      key = (ep_num, stp_num)
      if key not in seen_epochs:
        seen_epochs.add(key)
        epoch_rows.append((spd_str, stp_num, ep_num))
        render_monitor("Training")
      continue

    if NOISE_RE.search(line):
      continue

    if tb_lines_seen > 0:
      errors.append(line)
      tb_lines_seen = 0 if re.match(r"^[A-Za-z_]*(Error|Exception):", line) else (tb_lines_seen - 1)
      render_monitor("Error")
      continue

    if ERROR_START_RE.search(line):
      errors.append(line)
      if "Traceback" in line:
        tb_lines_seen = 10
      render_monitor("Error")
      continue

    if MILESTONE_RE.search(line):
      if not milestones or milestones[-1] != line:
        milestones.append(line)
        render_monitor("Running")
finally:
  if proc.poll() is None:
    proc.terminate()
  proc.wait()

# kills previously running processes
!fuser -k 6969/tcp
!fuser -k 8077/tcp
!fuser -k 9876/tcp
```
